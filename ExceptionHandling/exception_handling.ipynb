```python
f=None
try:
    f=open("dummy.txt","r")
    print(f.readlines())
except FileNotFoundError as fe:
    print("exception occured: ",fe)
finally:#finally block always executes weather an exception ocurred or not.
    if f:
       f.close()
    print("File closed")    
```

```python
#custom made exception
class ValueError(Exception):
    pass
balance=0
def deposit(amount):
    if amount<=0:
        raise ValueError("amount must be greater than zero")
    balance+=amount


try:
    deposit(0)
except ValueError as ve:
    print("exception occurred: ",ve)        
```
