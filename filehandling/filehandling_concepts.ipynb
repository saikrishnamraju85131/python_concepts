```python
#reading file content line by line

#opening the file in read mode
f=open("love.txt","r")

for line in f:
    print(line)

#closing the file
f.close()


```

```python
with open("love.txt","r") as f:
    for line in f:
        print(line)

#if we use "with",there is no need to close the file,it automatically closes the file.  
```

```python
#reading all the content in a file at a time using readlines() method.
with open("love.txt","r") as f:
    print(f.readlines()) #it returns list of lines as a result
```

```python
#writing data into a file
with open("memories.txt","w") as f:
    f.write("I love my college days")
```

```python
#writing multiple lines into a file
with open("memories.txt","w") as f:
    f.writelines(["I love spending time with my friends.","\neven though not doing great things in college life roaming in college is great experinece"])
```

```python
#problem with "w" mode is,it overrides the content in the file,instead of adding content at the end of file.

#In order to overcome these problem we use "a" mode.
```

```python
#working with "a" mode
with open("dummy.txt","a") as f:
    f.write("I love python\n")
```

```python
# writing multiple lines in dummy.txt file with "a" mode

with open("dummy.txt","a") as f:
    f.writelines(["I love Ai\n","I want to become top 1 per cent Ai engineer"])
```

```python
# in the above code we use write(),writelines() methods but content in the file is not overridden
#it appends at the end when we use "a" mode.

#printing the content line by line
with open("dummy.txt","r") as f:
    for line in f:
        print(line)
```

Lets us see failure scenarios in file handling

```python
#what happens when we try to read a content of a file,which does not exists

with open("user.txt","r") as f:
    for line in f:
        print(line)

#if we observe below,got FileNotFoundError,after getting these error program execution stops.        
```

```python
#creating a file with mode "x" 
 
with open("dummy.txt","x") as f:
    f.write("I love myself")

#whenever we try to create a file using mode "x",we make sure does not exist.
# if file already exists,then it raises FileExistsError,which causes program execution stops.    
```

calculating minimum,average,maximum scores of players by extracting data from csv file

```python
player_scores={}
with open("scores.csv","r") as f:
    for line in f:
        player,_,scores=line.split(",")
        if player not in player_scores:
            player_scores[player]=[int(scores.strip())]
        else:
            player_scores[player].append(int(scores.strip()))  

print(player_scores) 
```

```python
#calculating min,max,average of players

for player,score in player_scores.items():
    average=sum(score)/len(score)
    minimum=min(score)
    maximum=max(score)
    print(f"{player} average score is: {average},minimum_score is: {minimum},maximum_score is: {maximum}")
```
